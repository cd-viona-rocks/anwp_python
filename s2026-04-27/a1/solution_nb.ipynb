### Aufgabe 1: Analyse eines Kinosaals
Sie erhalten eine zweidimensionale Liste, die einen Kinosaal darstellt. Jede innere Liste entspricht einer Sitzreihe, und jeder Wert entspricht dem Alter der Person auf diesem Sitz. Ein Wert von 0 bedeutet, dass der Sitz leer ist.

```python
kinosaal = [
    [25, 34, 0, 18, 42],
    [0, 0, 51, 60, 33],
    [19, 22, 23, 0, 0],
    [45, 38, 0, 29, 31],
]
kinosaal
```

```python
chairs = []
empties = []
for r in kinosaal:
    one_row = [c for c in r if c != 0]
    chairs.append(one_row)
    empties.append(len(r) - len(chairs))
```

```python
# 1. anzahl belegten
anzahl = sum([len(t) for t in chairs])
print("die Gesamtanzahl der belegten Sitzen ist: ", anzahl)
```

```python
# 2. durchschnittsalter
sum_age = sum([sum(age) for age in chairs])
print("das Durchschnittsalter aller anwesenden Personen: ", sum_age // anzahl)
```

```python
# meinsten belegten Sitzen
# hier erstelle ich eine liste, um zu wissen, ob mehreren reihen geben, wo max anzahl ist
reihen = []
reihe_nr = 0
max_per_reihe = -1
for r in chairs:
    # statt index, loope ich over chairs und setze eine variable reihe_nr, welche schon mit 1 beginnt
    reihe_nr += 1
    if len(r) > max_per_reihe:
        max_per_reihe = len(r)
        # wenn großer, dann ersetze (bzg. resetiere) ich die liste reihe
        reihen = [reihe_nr]
    elif len(r) == max_per_reihe:
        # aber wenn mehreren gibt, dann füge ich einfach ein. 
        # wenn eine neue reihe kommt, wer mehr hat, dann ersetze ich einfach wie oben 
        reihen.append(reihe_nr)

# lösung
if len(reihen) == 1:
    print("die Reihe mit den meisten belegten Sitzen ist die ", reihen[0])
else:
    print("die Reihen mit den meisten belegten Sitzen sind die ", end="")
    print(*reihen, sep=", ", end="")
    print()

```

```python
# 4. sitze unter 21
# hier muss man wieder durch kino loopen

u21 = []
for x in range(len(kinosaal)):
    reihe = kinosaal[x]
    for y in range(len(reihe)):
        if reihe[y] != 0 and reihe[y] < 21:
            u21.append((x+1,y+1))

# lösung
if len(u21) == 1:
    print("die einzige Sitzposition, an denen jmd unter 21 sitzt, ist die ", u21[0])
else:
    print("die Sitzpositionen, an denen jmd unter 21 sitzen, sind die ", end="")
    print(*u21, sep=", ", end="")
    print()
```
