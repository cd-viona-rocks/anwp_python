### Aufgabe 2: Lagerverwaltung eines chaotischen Supermarkts
Ein Supermarkt hat sein Lager als zweidimensionale Liste organisiert. Jede innere Liste repräsentiert ein Regal, und jeder Wert repräsentiert die Anzahl eines Produkts.

```python
# lager -> (regal, anzahl prod)
lager = [ 
    [12, 4, 7, 0], 
    [3, 15, 0, 9], 
    [0, 2, 1, 8] 
    ]
```

Ihre Aufgaben:  

1. Berechnen Sie die Gesamtanzahl aller Produkte im Lager.  
2. Bestimmen Sie das Regal mit den meisten Produkten insgesamt.  
3. Erstellen Sie eine neue zweidimensionale Liste, in der alle Werte unter 5 auf 0 gesetzt werden. 4. Geben Sie alle Positionen aus, an denen ein Regal leer ist.


```python
# 1. anzahl aller produkte
anzahl = sum([sum(x) for x in lager])
print("die Gesamtanzahl aller Produkte im Lager ist", anzahl)
```

```python
# 2. anzahl pro regal
anz_regal = [sum(x) for x in lager]
max_anz = max(anz_regal)
regal = []
for x in range(len(anz_regal)):
    if anz_regal[x] == max_anz:
        regal.append(x+1)

if len(regal) == 1:
    print("das Regal mit den meisten Produkten insgesamt ist ", regal[0])
else:
    print("die Regale mit mehr Produkten insgesamt sind ", end="")
    print(*regal, sep=", ")
```

```python
new_lager = []
for r in lager:
    new_lager.append([x if x >= 5 else 0 for x in r])
print("die neue 2d Liste soll wie unten dargestellt werden")
for r in new_lager:
    print(r)
```

```python
# 4. nulle
pos = []
for x in range(len(lager)):
    r = lager[x]
    one_r = []
    for y in range(len(r)):
        if r[y] == 0:
            one_r.append(y)
    if len(one_r) > 0:
        pos.append((x+1, one_r))

print("alle Positionen aus, an denen ein Regal leer ist, sind:")
for x in pos:
    print(f"Regal {x[0]} -> Position{"en" if len(x[1]) > 1 else ""}: ", end="")
    print(*[e+1 for e in x[1]], sep=", ")
```
