### Aufgabe 3: Punktevergabe in einem sehr unfairen Wettbewerb

Eine Jury bewertet Teilnehmer in mehreren Kategorien. Die Bewertungen sind in einer zweidimensionalen Liste gespeichert. Jede innere Liste entspricht einem Teilnehmer, und jeder Wert ist die Punktzahl einer Kategorie.

```python
punkte = [ 
    [8, 7, 9, 6], 
    [5, 6, 5, 4], 
    [9, 9, 10, 8], 
    [3, 4, 2, 5] 
]
```

Ihre Aufgaben: 

1. Berechnen Sie die Gesamtpunktzahl jedes Teilnehmers. 
2. Finden Sie den Teilnehmer mit der höchsten Gesamtpunktzahl. 
3. Berechnen Sie den Durchschnitt jeder Kategorie. 
4. Erstellen Sie eine neue Liste, die nur Teilnehmer enthält, deren Gesamtpunktzahl über 25 liegt.

```python
# 1. ges.anzahl
punkt_tn = [(x+1, sum(punkte[x])) for x in range(len(punkte))]
print("die Gesamtpunktzahl jedes Teilnehmers lautet: ")
for x in punkt_tn:
    print(f"TN {x[0]} -> {x[1]} Punkt{"e" if x[1] > 1 else ""}")
```

```python
# 2. tn mit höchsten ges P
punkt_tn.sort(key=lambda x: x[1], reverse=True)
tn = punkt_tn[0]
print(f"der Teilnehmer mit der höchsten Gesamtpunktzahl ist der TN {tn[0]} mit {tn[1]} Punkt{"e" if x[1] > 1 else ""}")
```

```python
# 3. avg / cat
cat = [0] * 4
anz_cat = len(punkte[0])
for tn in punkte:
    for x in range(anz_cat):
        cat[x] += tn[x]

print("der Durchschnitt jeder Kategorie ist: ")
for x in range(anz_cat):
    cat[x] /= anz_cat
    print(f"Kategorie {x+1} -> {cat[x]:.2f} Punkte")
```

```python
# 4. liste tn > 25
punkt_tn25 = list(filter(lambda x: x[1] > 25, punkt_tn))
tn_list = [tn[0] for tn in punkt_tn25]
list_tn25 = [punkte[x] for x in range(len(punkte)) if x in tn_list]
print("die neue Liste mit TN, deren Ges.P über 25 steht, lautet: ")
for x in list_tn25:
    print(x)
```
